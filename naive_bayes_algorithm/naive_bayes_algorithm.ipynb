```python
import pandas as pd
```

```python
df=pd.read_csv("play_tennis.csv")
df
```

```python
target = df['play']
attributes = df.drop(columns=['day', 'play'])
```

```python
def prior(target,label):
    return sum(target==label)/len(target)
```

```python
def conditional(feature,value,target,label):
    count_label=sum(target==label)
    count_feature=sum((target==label)&(feature==value))
    if(count_label>0):
        return count_feature/count_label
    else:
        return 0
```

```python
def posterior(test):
    classes=['Yes','No']
    posteriors=[]
    for label in classes:
        prior_prob=prior(target,label)
        prob=prior_prob
        for feature_name in attributes.columns:
            prob*=conditional(df[feature_name],test[feature_name],target,label)
        print(f"Posterior probability for class '{label}':{prob}")
        posteriors.append(prob)
    return posteriors
```
