```python
import numpy as np
import pandas as pd
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import classification_report,confusion_matrix,ConfusionMatrixDisplay
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("IRIS.csv")
df
```

```python
iris=datasets.load_iris()
iris
```

```python
X=iris.data[:,:2]
y=iris.target
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
iris
```

```python
svc=SVC(kernel='linear',C=1)
svc.fit(X_train,y_train)
```

```python
y_pred=svc.predict(X_test)
```

```python
print("Confusion Matrix")
print(confusion_matrix(y_test,y_pred))
ConfusionMatrixDisplay.from_estimator(svc,X_test,y_test)
```

```python
print("Classification Report:")
print(classification_report(y_test,y_pred))
```

```python
plt.scatter(X[:,0],X[:,1],c=y,edgecolor='k')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('SVM Algorithm')
plt.show()
```

```python
plt.scatter(X[:,0],X[:,1],c=y,edgecolors='k',label='Data Points')
plt.scatter(svc.support_vectors_[:,0],svc.support_vectors_[:,1],s=100,facecolors='none',edgecolors='r',label='Support Vectors')
w=svc.coef_[0]
b=svc.intercept_[0]
x_range=np.linspace(X[:,0].min(),X[:,0].max(),100)
y_range=-(w[0]/w[1])*x_range-b/w[1]
plt.plot(x_range,y_range,'g--',label='Max Margin Hyperplane')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.title('SVM Algorithm Visualization')
plt.show()
```

```python
x_min,x_max=X[:,0].min()-1,X[:,0].max()+1
y_min,y_max=X[:,1].min()-1,X[:,1].max()+1
xx,yy=np.meshgrid(np.arange(x_min,x_max,0.01),np.arange(y_min,y_max,0.01))
Z=svc.predict(np.c_[xx.ravel(),yy.ravel()])
Z=Z.reshape(xx.shape)
plt.contourf(xx,yy,Z,alpha=0.8)
plt.scatter(X[:,0],X[:,1],c=y,edgecolors='k',marker='o')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('SVM Decision Boundaries for Iris Dataset')
plt.show()
```
